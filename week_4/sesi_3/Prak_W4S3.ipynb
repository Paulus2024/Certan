```python
import numpy as np # imbil alat dari library numpy
import pandas as pd # imbil alat dari library pandas
from sklearn.datasets import load_iris # dari sklearn, subdataset ambil load_iris

# Load dataset Iris dari scikit-learn
iris = load_iris() # buat variable iris untuk menampung dataset load_iris()
```

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names) # buat table, ambil data dari iris.data, berikan nama nama columns sesua dari iris.feature_names
df["Species"] = iris.target_names[iris.target]# buat columns baru dan setiap kolom berikan nama sesuai dengan speciesnya
```

```python
# Definisikan data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names) # buat table, ambil data dari iris.data, berikan nama nama columns sesua dari iris.feature_names
df["Species"] = iris.target_names[iris.target] # buat columns baru dan setiap kolom berikan nama sesuai dengan speciesnya

# Definisikan x (fitur) dan y (target/label)
X = df.drop(columns=["Species"]) # ambil semua kolom kecualu species karena akan di jadikan soal tebakan
y = df["Species"] # ambil kolom species untuk hasil tebakan (jawaban)

print(X) # tampilkan X ke layar
```

```python
y # tampilkan jawaban hasil tebakan
```

```python
from sklearn.preprocessing import LabelEncoder # ambil data dari sklearn, masuk ke sub bab prepocessing dan ambil data LabelEncoder
lab = LabelEncoder() # buat variable lab dan simpan data dari LabelEncoder()
y = lab.fit_transform(y)
y
```

```python
from sklearn.model_selection import train_test_split # ambil data dari sklearn, masuk ke sub bab model_selection dan ambil data train_test_split

# X_train, X_test = train_test_split(X,test_size=0.2, random_state=0)
# y_train, y_test = train_test_split(y,test_size=0.2, random_state=0)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0) # bagi data menjadi 2 bagian, 80% untuk data latih dan 20% untuk data uji, random_state=0 untuk mengacak data
```

```python
from sklearn.naive_bayes import GaussianNB # ambil data dari sklearn, masuk ke sub bab naive_bayes dan ambil data GaussianNB

model = GaussianNB() # buat variable model untuk menampung data dari GaussianNB()
NB_model = model.fit(X_train, y_train) # latih model dengan data latih X_train dan y_train
```

```python
result = NB_model.predict(X_test) #
result #
```

```python
from sklearn.metrics import accuracy_score # ambil data dari sklearn, masuk ke sub bab metrics dan ambil data accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100) # tampilkan akurasi dari data uji, dengan membandingkan data uji dengan hasil tebakan
```

```python
# Prediksi inputan baru
# inpput_data = pd.DataFrame([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]] # buat variable input_data untuk menampung data baru yang akan di prediksi, data baru ini harus memiliki jumlah kolom yang sama dengan data latih
result = NB_model.predict(input_data) # eksekusi model untuk memprediksi data baru yang telah di masukkan ke dalam variable input_data
prediction_label = lab.inverse_transform(result)# ubah hasil prediksi dari angka menjadi label yang sesuai dengan speciesnya
print('Hasil prediksi untuk input baru adalah:', prediction_label) # cetak hasil prediksi ke layar
```
