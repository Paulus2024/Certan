```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-1.5, 1.5, 400)
y = np.linspace(-1.5, 1.5, 400)
X, Y = np.meshgrid(x, y)
```

```python
F = (
    X**6
    + 3 * X**4 * Y**2
    - 3 * X**4
    + 3 * X**2 * Y**4
    - X**2 * Y**3
    - 6 * X**2 * Y**2
    + 3 * X**2
    + Y**6
    - 3 * Y**4
    + 3 * Y**2
    - 1
)
```

```python
plt.contour(X, Y, F, levels=[0])
plt.show()
```
