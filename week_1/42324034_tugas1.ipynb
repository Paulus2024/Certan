```python
import numpy as np
```

```python
matriks = np.random.randint(1, 100, size=(5, 5))
```

```python
nilai_max = matriks.max()
nilai_min = matriks.min()
nilai_rata = matriks.mean()
```

```python
print("Matriks 5x5:")
print(matriks)
print("\nNilai Maksimum :", nilai_max)
print("Nilai Minimum  :", nilai_min)
print("Nilai Rata-rata:", nilai_rata)
```
