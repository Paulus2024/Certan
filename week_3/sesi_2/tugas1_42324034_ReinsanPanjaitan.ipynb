```python
import pandas as pd
df = pd.read_csv('rounded_hours_student_scores.csv')
print(df.head())
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values
```

```python
import matplotlib.pyplot as paulus
paulus.scatter(X, y, color = "red")
paulus.xlabel('Waktu Belajar (Jam)')
paulus.ylabel('Score')
paulus.show()
```

```python
# from sklearn.model_Selection import train_test_split
# X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
a = regressor.intercept_
b = regressor.coef_[0]
print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"Persamaan garis: y = {a:.2f} + {b:.2f}x")
```

```python
y_pred = regressor.predict(X_test)
print(y_pred)

from sklearn.metrics import r2_score
score = r2_score(y_test, y_pred)
print("Nilai R-Square:", score)
```

```python
paulus.scatter(X_test, y_test, color = 'red')
paulus.plot(X_train, regressor.predict(X_train), color = 'blue')
paulus.title('Score vs Waktu Belajar (Test set)')
paulus.xlabel('Waktu Belajar (Jam)')
paulus.ylabel('Score')
paulus.show()
```
